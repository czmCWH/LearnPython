# 1、数据排序
在进行数据排序时，有两种排序方式，分别是：升序和降序。而基于 Pandas 进行数据排序时，是**可以按照多个列进行排序**。

多列排序：进行多列排序时，会先按照第一列进行排序，第一列的值相同时，才会按照第二列进行排序，以此类推。

```python
import pandas as pd

# 读取数据，只读取前10行数据
df = pd.read_csv('./data/sales.csv', nrows=10)
df
```

```python
print('1、根据 销售数量 降序排序：')
# ascending 默认为True，表示升序排序，设置为False表示降序排序。
print(df.sort_values(by='销售数量', ascending=False))

print('--------------------------------------------------')

print('2、根据 单价 生序排序：')
print(df.sort_values(by='单价', ascending=True))

print('--------------------------------------------------')

print('3、根据 单价 升序排序，价格一样时，根据 销售数量 降序排序：')
print(df.sort_values(by=['单价', '销售数量'], ascending=[True, False]))

```

# 2、数据分组
分组操作就是把数据按照某个特征分成不同的组，然后对每个组分别进行统计计算。

数据分组的目的是，对分组后的数据进行统计分析，常用的操作有：
  - `sum()`：计算分组后某一列数值的总和；
  - `count()`：统计有效数量，统计分组后某一列非空（Non-NaN）数据的个数（若需包含空值统计总行数，请用 size()）；
  - `min()`、`min()`：获取分组后的某一列最大或最小值（不仅支持数值，也支持字符串的字典序和时间日期的先后顺序）
  - `mean()`：计算数值列的算术平均值；
  - `median()`：计算数值列的中位数（能有效剔除极值影响）；
  - 聚合操作：`agg()`，一次性完成多个操作；

⚠️，在进行统计操作时，缺失值不参与统计。

```python
import pandas as pd

# 读取数据，只读取前20行数据
df = pd.read_csv('./data/sales.csv', nrows=20)
df
```

```python
print('👉 1、根据 产品类别 进行分组，统计各个类别的 订单数量：')
# ⚠️，.count() 统计的是非空（Non-NaN）单元格数量。如果某行的 '订单号' 为空，该行在计数时会被直接跳过（不增加计数），绝非“当作 0 处理”。
print(df.groupby('产品类别')['订单号'].count())

print('--------------------------------------------------')
print('👉 2、根据 产品类别 进行分组，统计各个类别的 销售数量 之和：')
print(df.groupby('产品类别')['销售数量'].sum())
print('--------------------------------------------------')
print('👉 3、根据 产品类别 进行分组，统计各个类别的 销售金额 之和：')
# 1、先计算销售金额
df['销售金额'] = df['销售数量'] * df['单价']
# 2、根据 产品类别 进行分组，统计各个类别的 销售金额 之和
print(df.groupby('产品类别')['销售金额'].sum())
print('--------------------------------------------------')
print('👉 4、根据 产品类别 进行分组，统计各个类别的最低商品 单价：')
print(df.groupby('产品类别')['单价'].min())
print('--------------------------------------------------')
print('👉 5、根据 产品类别 进行分组，统计各个类别的最高商品 单价：')
print(df.groupby('产品类别')['单价'].max())
print('--------------------------------------------------')
print('👉 6、根据 产品类别 进行分组，统计各个类别的平均商品 单价：')
print(df.groupby('产品类别')['单价'].mean())
print('--------------------------------------------------')
print('👉 7、根据 产品类别 进行分组，统计各个类别的商品 平均单价、最高单价、最低单价：')
print(df.groupby('产品类别')['单价'].agg(['mean', 'max', 'min']))
print('--------------------------------------------------')
print('👉 8、根据 产品类别 进行分组，统计各个类别的商品 销售数量 之和，销售金额 之和，平均单价')
print(df.groupby('产品类别').agg({'销售数量': 'sum', '销售金额': 'sum', '单价': 'mean'}))
print(df.groupby('产品类别').agg({'销售数量': 'sum', '销售金额': 'sum', '单价': 'mean'}).to_dict())
```
