# 👉 1、数据查看、选择、过滤

数据从文件中加载出来后，数据量很大，不方便直接查看数据内容。因此，需要查看数据的内容，选择需要的数据，过滤不需要的数据。

* `查看`：因此，需要查看数据的前几行，或者查看数据的后几行，或者查看数据的整体情况。
Pandas 提供了如下查看方法：
  - df.head(n)，查看数据的前 n 行；
  - df.tail(n)，查看数据的后 n 行；
  - df.describe()，查看数据中每一列的描述性统计信息，如：均值、标准差、最大值、最小值等；
  - df.info()，查看数据的整体信息，如：列名、非空值数量、列数据类型等；
  - df.shape，查看数据的维度，返回一个元组(行数, 列数)；
  - df.columns，查看数据的列名；
  
* `选择`：是指从数据中选取一部分数据，如：选择列、选择行。选择数据时，可以使用如下方法：
  - `df[‘列名’] 或 df.列名`，根据列名选择某一列数据；
  - `df[['列名1', '列名2']] 或 df.[['列名1', '列名2']]`，根据多个列名选择多列数据，返回值是一个 DataFrame；
  - df.iloc[start:stop:step:]，根据 `行号` 选择数据，即进行数据切片。选择行号范围 `不` 包含 stop；
  - df.loc[start:stop:step:]，根据 `行索引标签` 选择数据，即进行数据切片。选择行标签范围包含 stop；
  - df.loc[:, []]，根据行标签和列名选择数据；

* `过滤`：是指从数据中根据条件筛选一部分数据。过滤数据时，可以使用如下方法：
  - `df[df["列名"] >= 0]`，筛选出指定列中数值`大于或等于 0` 的所有行。
  - `df[df["列名"].isin([值1, 值2, ...])]`，筛选出指定列的值`存在于给定列表`中的所有行。
  - `df[df["列名"].str.contains("字符串")]`，筛选出指定列的文本中 `包含特定子字符串` 的所有行。
  - `df[df["列名"].between(值1, 值2)]`, 筛选出指定列的值在 `[值1, 值2] 闭区间`之内的所有行（包含边界值1和值2）。
  - 组合多个条件过滤数据，可以使用 `&` 或 `|` 运算符；

```python
import pandas as pd

# 读取csv文件
# 参数 usecols 来指定需要读取的列。
# 参数 index_col 来指定某一列为索引标签。
df = pd.read_csv('./data/sales.csv', usecols=['订单号', '产品类别', '产品名称', '销售数量', '单价', '支付方式'], index_col='订单号')

# ⚠️ 配置项，显示所有数据（默认打印获取 Pandas 的数据时，数据太多会被 Jupyter 折叠显示，通过如下配置项，可以使数据全部展示）
# pd.set_option('display.max_columns', None)
```

```python
# 👉 2、数据查看
print("df.head(5) 查看数据的前5行：")
print(df.head(5))
print("----------------------------------")
print("df.tail(5) 查看数据的后5行：")
print(df.tail(5))
print("----------------------------------")
print("df.describe() 查看数据中每一列的统计信息：")
print(df.describe())
print("----------------------------------")
print("df.info() 查看数据的整体信息：")
print(df.info())
print("----------------------------------")
print("df.shape 查看数据的维度：")
print(df.shape)
print("----------------------------------")
print("df.columns 查看数据的列名：")
print(df.columns.tolist())
print("----------------------------------")
```

```python
# 👉 3、数据选择：选择列、选择行
print("df['产品名称']，选择某一列的数据：")
print(df['产品名称'])
print("----------------------------------")
print("df[['产品类别', '销售数量']]，选择多列的数据：")
print(df[['产品类别', '销售数量']])
print("----------------------------------")
print("df.iloc[0:5]，根据行号选择数据，即选择前5行数据：")
print(df.iloc[0:5])
print("----------------------------------")
print("df.loc[6805677496:5804178086]，根据行索引标签选择数据：")
print(df.loc[6805677496:5804178086:3])
print("----------------------------------")
print("df.loc[6805677496:5804178086, ['产品类别', '销售数量']]，根据行索引标签选择数据，并指定列：")
print(df.loc[6805677496:5804178086, ['产品类别', '销售数量']])
```

```python
# 👉 4、数据过滤
# 1、获取 销售数量 >= 10 的订单数据
print("df[df['销售数量'] >= 10]，根据列名和条件过滤数据：")
print(df[df['销售数量'] >= 10])
print("----------------------------------")

# 2、获取 产品类别 为 食品 或 图书 的订单数据
print("df[df['产品类别'].isin(['食品', '图书'])]，根据列名和条件过滤数据：")
print(df[df['产品类别'].isin(['食品', '图书'])])

print("----------------------------------")

# 3、获取单价在 100～200 元之间的订单数据
print("df[df['单价'].between(100, 200)]，根据列名和条件过滤数据：")
print(df[df['单价'].between(100, 200)])
print("----------------------------------")

# 4、获取 销售数量 >= 8，并且 单价 >= 100 的订单数据
print("df[(df['销售数量'] >= 8) & (df['单价'] >= 100)]，根据列名和条件过滤数据：")
print(df[(df['销售数量'] >= 8) & (df['单价'] >= 100)])
print("----------------------------------")

# 5、获取产品类别为 服装/食品，并且支付方式为：支付宝/微信支付 的订单数据
print("df[(df['产品类别'].isin(['服装', '食品'])) & (df['支付方式'].isin(['支付宝', '微信支付']))]，根据列名和条件过滤数据：")
print(df[(df['产品类别'].isin(['服装', '食品'])) & (df['支付方式'].isin(['支付宝', '微信支付']))])
print("----------------------------------")
```
