# 👉 1、Pandas 中数据的读取与写入

基于 Pandas 库中提供的API，可以很方便的各类数据文件(csv、Excel、数据库、网络数据等)的读取和写入。

处理流程：
  - 读取数据，通过 .read_csv、read_excel 等方法读取数据到 DataFrame；
  - 处理 DataFrame 中的数据：数据清洗处理、数据分析、数据可视化；
  - 将结果写入到文件中，通过 .to_csv()、.to_excel() 等方法写入到文件中；

* 注意，通过 .read_excel() 方法第一次读取 Excel 文件时，可能会报错，这是因为 Pandas 没有安装对应的 Excel 引擎。直接根据提示安装即可。

```python
import pandas as pd

# 读取csv文件
# 通过参数 usecols 来指定需要读取的列。
df = pd.read_csv('./data/sales.csv', usecols=['订单号', '产品类别', '产品名称', '销售数量', '单价'])
print(df)

# 处理数据：新增一列：销售额
df['销售额（元）'] = df['销售数量'] * df['单价']
print(df)

# 保存数据到csv文件
df.to_csv('./data/sales_new.csv', index=False)
```
