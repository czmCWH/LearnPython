```python
# 👉 1、Pandas 入门
import pandas as pd

# 1、创建一个 DataFrame 对象，即创建数据集（学习成绩信息）
df = pd.DataFrame([
  {"姓名": "小王", "语文": 90, "数学": 70, "英语": 87},
  {"姓名": "小李", "语文": 80, "数学": 90, "英语": 78},
  {"姓名": "小张", "语文": 85, "数学": 80, "英语": 90},
  {"姓名": "小赵", "语文": 75, "数学": 85, "英语": 80},
  {"姓名": "小刘", "语文": 95, "数学": 70, "英语": 88},
  {"姓名": "小陈", "语文": 88, "数学": 75, "英语": 85},
  {"姓名": "小孙", "语文": 90, "数学": 70, "英语": 87}
])
# 2、查看数据集
print(df)

# 3、分析数据集，各科成绩的最高分、最低分、平均分
# df['语文'] 就是一个 Series 对象，可以直接调用 max()、min() 和 mean() 方法来获取最高分、最低分和平均分。
print(f"语文最高分：{df['语文'].max()}，最低分：{df['语文'].min()}，平均分：{df['语文'].mean():.2f}")
print(f"数学最高分：{df['数学'].max()}，最低分：{df['数学'].min()}，平均分：{df['数学'].mean():.2f}")
print(f"英语最高分：{df['英语'].max()}，最低分：{df['英语'].min()}，平均分：{df['英语'].mean():.2f}")

# ⚠️ 总结：如上基于 DataFrame 对象，我们可以轻松地获取到各科成绩的最高分、最低分、平均分等信息。
```

```python
# 👉 2、构建 DataFrame 有很多方式，主要方式如下：
import pandas as pd

df1 = pd.DataFrame({
  '姓名': ['小明', '小红', '小刚', '小花'],
  '语文': [90, 80, 85, 95],
  '数学': [70, 90, 80, 85],
  '英语': [87, 78, 90, 80]
})

df2 = pd.DataFrame([
  ('小明', 90, 70, 87),
  ('小红', 80, 90, 78),
  ('小刚', 85, 80, 90),
], columns=['姓名', '语文', '数学', '英语'])

df3 = pd.DataFrame([
  ['张三', 90, 70, 87],
  ['李四', 80, 90, 78],
  ['王五', 85, 80, 90],
], columns=['姓名', '语文', '数学', '英语'], index=['a', 'b', 'c'])

print(df3)
```

```python
# 👉 3、DataFrame 对象的属性

df3 = pd.DataFrame([
  ['张三', 90, 70, 87],
  ['李四', 80, 90, 78],
  ['王五', 85, 80, 90],
], columns=['姓名', '语文', '数学', '英语'], index=['a', 'b', 'c'])

print(df3)
print('------------------------------')
print('df3.index 每一行的索引：')
print(df3.index)
print('------------------------------')
print('df3.column 每一列的索引：')
print(df3.columns)
print('------------------------------')
print('df3.values.tolist() 每一行的值：')
print(df3.values.tolist())
print('------------------------------')
print('df3.dtypes 每一列数据的类型（⚠️，此处返回的类型不是 Python 中的类型，是 Pandas 内置的类型）：')
print(df3.dtypes)
print('------------------------------')
print('df3.size 单元格的数量：')
print(df3.size)
print('------------------------------')
print('df3.shape 数据集的数据维度：几行、几列：')
print(df3.shape)
```

```python
# 👉 4、构建 Series 对象（列）
import pandas as pd

s = pd.Series([1, 2, 3, 4, 5])    # 未指定索引，默认从 0 开始
print("构建默认索引的 Series 对象：")
print(s)
print('------------------------------')
s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])    # 指定索引
print("构建指定索引的 Series 对象：")
print(s)
print('------------------------------')
s = pd.Series({'a': 11, 'b': 22, 'c': 33, 'd': 44, 'e': 55})    # 使用字典创建，索引默认为字典的键，值作为元素
print("使用字典创建的 Series 对象：")
print(s)
print('------------------------------')
# 通过列名获取 DataFrame 中的 Series 对象，DataFrame 中的每一列都是一个 Series 对象。
s = df1['姓名']    # 这里的 df1 是 DataFrame，'姓名' 是列名
print("通过列名获取 DataFrame 中的 Series 对象：")
print(s)  
```

```python
# 👉 5、Series 对象的属性
print("s.index.tolist() 表示 Series 对象的 索引：")
print(s.index.tolist())
print('------------------------------')
print("s.values.tolist()，表示 Series 对象所对应一列的数据：")
print(s.values.tolist())
print('------------------------------')
print("s.dtype，表示 Series 对象中整个数据的 数据类型：")
print(s.dtype) # ⚠️，这里获取的数据类型不是 Python 中的类型，是 Pandas 内置的类型
print('------------------------------')
print("s.size，表示 Series 对象的一列中数据的 个数：")
print(s.size)
print('------------------------------')
print("s.shape，表示 Series 对象的 数据维度，即：（行, ）。只输出行，列是固定的只有1列：")
print(s.shape)
```
